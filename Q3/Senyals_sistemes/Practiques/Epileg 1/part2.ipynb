```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve, convolve2d, correlate2d
from math import sqrt
```

```python
def find_by_correlation (img: np.ndarray, noise: np.ndarray) -> tuple[np.intp, np.intp]:
    """Finds the image amongst the noise. Returns the first pixel of the found image."""
    correlation = correlate2d(noise, img, mode = "same")
    x, y = np.unravel_index(np.argmax(correlation), correlation.shape)
    return (x, y)
```

```python
# Finding SIS in a noisy image
noise1 = plt.imread("find_sis_n7.png")
SIS = plt.imread("logo_sis.png")
i1, j1 = find_by_correlation(SIS, noise1)
print(i1, j1)
plt.figure(7)
plt.imshow(noise1[i1:i1+5, j1:j1+6], cmap = 'gray')


plt.show()
```

```python
noise2 = plt.imread("find_sis_n13.png")
SIS = plt.imread("logo_sis.png")
i2, j2 = find_by_correlation(SIS, noise2)
print(i2, j2)
plt.figure(8)
plt.imshow(noise2[i2-4:i2+5, j2-5:j2+6], cmap = 'gray')
```
